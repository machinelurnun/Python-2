```python
# Just as def defines a function but does not call it, 
# class defines an object but does not instantiate one
class Point:
    '''
    Represents a point in 2D space

    attributes: x (float), y (float)
    '''

    # Constructor always requires self as the first argument, followed by whatever attributes you need
    # The attributes here have been assigned default values
    def __init__(self, input_x = 0, input_y = 0):
        self.x = input_x
        self.y = input_y
```

```python
# Instantiating a Point object
exPoint = Point(5, 10)
print(exPoint.x, exPoint.y)
```

```python
origin = Point()
print(origin.x, origin.y)
```

```python
# Here is a function that expects an object as input.
def dist_from_origin(inputPoint):
    '''

    :param inputPoint: Point object
    :return: Distance between inputPoint and origin (0, 0)
    '''

    import math

    x_coord = inputPoint.x
    y_coord = inputPoint.y
    dist = math.sqrt(x_coord ** 2 + y_coord ** 2)
    return dist
```

```python
exPoint2 = Point(3,4)
dist_from_origin(exPoint2)
```

```python
def get_center_Point(inputRect):
    '''

    :param inputRect: Rectangle object
    :return: Point object corresponding to the center of inputRect
    '''

    # PRACTICE: Fill in the code that computes and returns the center of the rectangle
    # as a Point object
    corner_x = inputRect.corner.x
    corner_y = inputRect.corner.y
    rect_width = inputRect.width
    rect_height = inputRect.height
    # formula to find horizontal and vertical midpoint coordinates
    horizontal_midpoint = (corner_x + rect_width/2)
    vertical_midpoint = (corner_y + rect_height/2)
    # finds midpoint coordinates x (horizontal_midpoint) and y (vertical_midpoint).
    midpoint = Point(horizontal_midpoint,vertical_midpoint)
    return midpoint
```

```python
# You can overwrite attributes directly, but it is bad practice.
# We'll see a better way soon.
pointA = Point(0, 0)
pointA.x = 5
print(pointA.x, pointA.y)
```

```python
# Be careful about accidentally aliasing objects.
# Python does not create a new copy when you only use the = operator.
pointB = pointA

# The x and y of pointB will overwrite the values already stored in pointA
# because pointA and pointB are aliases for the same data in memory.
pointB.x = 50

print(pointA.x ,pointA.y)
print(pointB.x, pointB.y)
```

```python
# Use the copy function from the copy library to make true copies of objects
import copy # import the copy library

pointA = Point(5, 0)
pointB = copy.copy(pointA) # make a copy of pointA
pointB.x = 50

print(pointA.x, pointA.y)
print(pointB.x, pointB.y)
```
