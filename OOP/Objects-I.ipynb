```python
class Point:
    '''
    Represents a point in 2D space

    attributes: x (float), y (float)
    '''

    # Constructor always requires self as the first argument, followed by whatever attributes you need
    # The attributes here have been assigned default values
    def __init__(self, input_x = 0, input_y = 0):
        self.x = input_x
        self.y = input_y
```

```python
# One purpose of classes is to bundle together (encapsulate) related data so it can be
# reused in other classes/functions.
class Rectangle:
    '''
    Represents a rectangle with lower left corner at corner and specified height and width.

    attributes: corner (Point object), height (float), width (float)
    '''

    def __init__(self, input_corner = Point(), input_height = 0, input_width = 0):
        self.corner = input_corner
        self.height = input_height
        self.width = input_width
```

```python
rect2 = Rectangle(Point(1,2),3,4)
```

```python
def get_corners(inputREC):

#width = x axis
#height = y axis
    initial_corner = (inputREC.corner.x, inputREC.corner.y)
    bottom_right = (inputREC.corner.x + inputREC.width,inputREC.corner.y)
    upper_left = (inputREC.corner.x, inputREC.corner.y + inputREC.height)
    upper_right = (inputREC.corner.x + inputREC.width , inputREC.corner.y + inputREC.height)
    object_list = [initial_corner, bottom_right, upper_left, upper_right]
    return object_list
```

```python
get_corners(rect2)
```

```python
def distance(Point1, Point2):
    import math
    formula_x = Point2.x - Point1.x
    formula_y = Point2.y - Point1.y
    squared_x = formula_x^2
    squared_y = formula_y^2
    sqrt = math.sqrt(squared_x + squared_y)
    return sqrt
```

```python
class Circle():

    def __init__(self, center, radius):
        self.center = Point()
        self.radius = radius
```

```python
def point_in_circle(circle, point):
    distance_from_cent = distance(circle.center, point)
    if circle.radius >= distance_from_cent:
        return True
    else:
        return False
```

```python
def rect_in_circle(circle,rectangle):
    for corner in get_corners(rectangle):
        if not point_in_circle(circle,Point(corner[0], corner[1])):
            return False
    return True
```

```python
rec1 = Rectangle(Point(1,2),3,4)
circle1 = Circle(Point(1,2),1)
print(rect_in_circle(circle1, rec1))
```

```python
rec2 = Rectangle(Point(0,0),1,1)
Circle2 = Circle(Point(1,1),5)
print(rect_in_circle(Circle2, rec2))
```
