```python
def selection_sort(nList, asc=True):
    if asc:
        for iteration in range(len(nList)):
            min_loc = iteration
            for i in range(min_loc+1,len(nList)):
                if nList[i] < nList[min_loc]:
                    min_loc = i
            if min_loc != iteration:
                (nList[iteration],nList[min_loc]) = (nList[min_loc],nList[iteration])
    
    else:
        for iteration in range(len(nList)-1):
            min_loc = iteration
            for i in range(min_loc+1,len(nList)):
                if nList[i] > nList[min_loc]:
                    min_loc = i
            if min_loc != iteration:
                (nList[iteration],nList[min_loc]) = (nList[min_loc],nList[iteration])

def insertion_sort(nList, asc=True):
    if asc:
        for i in range(1,len(nList)):
            j = i
            while j > 0 and nList[j-1] > nList[j]:
                (nList[j],nList[j-1]) = (nList[j-1],nList[j])
                j = j-1
    else:
         for i in range(1,len(nList)):
                    j = i
                    while j > 0 and nList[j-1] < nList[j]:
                        (nList[j],nList[j-1]) = (nList[j-1],nList[j])
                        j = j-1
```

```python
test_list = [5, 3, 4, 2, 1]
insertion_sort(test_list)
print(test_list)
```
