```python
def overlap(s1, s2):
    overlap2 = 0
    for i in range(len(s1)):
        if s1[-i:] == s2[0:i]:
            overlap2 = i
    return overlap2

def merge(s1, s2, pOver):
    return s1 + s2[pOver:]


def shortest_superstring(stringSet):
    string_list = list(stringSet)

    while len(string_list) > 1:
        #print(string_list)
        largest_overlap = 0
        best_strings = (0,0)
        best_merged_string = ' '

        for a in range(len(string_list)):
            for b in range(len(string_list)):
                if a != b:
                    string1, string2 = string_list[a], string_list[b]
                    over_lap = overlap(string1, string2)

                    if over_lap > largest_overlap:
                        largest_overlap = over_lap
                        best_strings = (a,b)
                        best_merged_string = merge(string1, string2, largest_overlap)
        if largest_overlap == 0:
            best_merged_string = string_list[0] + string_list[1]
        modified_strings = list()

        for c in range(len(string_list)):
            if c != best_strings[0] and c != best_strings[1]:
                modified_strings.append(string_list[c])

        modified_strings.append(best_merged_string)
        string_list = modified_strings
    return string_list[0]
```

```python
print(shortest_superstring({'ABCDEF', 'BCDEFX', 'XCDEFY', 'DEFYZ'}))
shortest_superstring({'UV', 'VW', 'XY', 'YZ'})
```
