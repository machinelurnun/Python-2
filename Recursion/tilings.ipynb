```python
solutions = list()
def generate_tilings(partial_tiling, length):
    current_sum = sum(partial_tiling)
    if current_sum == length:
            global solutions
            solutions.append(partial_tiling)

    elif current_sum < length:          
        for i in range(1, 5):
             list_copy = partial_tiling.copy()
             list_copy.append(i)
             generate_tilings(list_copy,length)
    return solutions
        
          

                        

                      
```

```python
generate_tilings([], 4)
```
