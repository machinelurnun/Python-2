```python
def merge(a,b):
    c = list()
    while len(a) > 0 and len(b) > 0:
            
            if a[0] > b[0]:
                b_num = b.pop(0)
                c.append(b_num)
            else:
                a_num = a.pop(0)
                c.append(a_num)
                
    while len(a) > 0:
         a_num2 = a.pop(0)
         c.append(a_num2)

    while len(b) > 0:
         b_num2 = b.pop(0)
         c.append(b_num2)
    return c

def mergesort(a):
     if len(a) == 1:
          return a
     
     half_length = len(a) // 2
     arrayOne = a[:half_length]
     arrayTwo = a[half_length:]

     arrayOne = mergesort(arrayOne)
     arrayTwo = mergesort(arrayTwo)

     return merge(arrayOne,arrayTwo)

```

```python
mergesort([6, 5, 5, 7, 4, 8, 3, 5, 1, 2, 9])
```

```python
def selection_recursive(intList):
    if len(intList) <= 1:
        return intList
    
    largest = max(intList)

    remaining = intList.copy()
    remaining.remove(largest)

    result = selection_recursive(remaining)
    result.append(largest)
    return result


    
```

```python
selection_recursive([6, 5, 5, 7, 4, 8, 3, 5, 1, 2, 9])
```
