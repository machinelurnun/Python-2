```python
solutions = list()
def generate_tilings(partial_tiling, length):
    current_sum = sum(partial_tiling)
    if current_sum == length:
            global solutions
            solutions.append(partial_tiling)

    elif current_sum < length:          
        for i in range(1, 5):
             list_copy = partial_tiling.copy()
             list_copy.append(i)
             generate_tilings(list_copy,length)
    return solutions     
```

```python
generate_tilings([], 4)
```

```python
solutions = list()
def subset_sum(partial_sum, remaining_ints, target):
    current_sum = sum(partial_sum)
    if current_sum == target:
        solutions.append(partial_sum)
        return
    elif current_sum > target:
        return
    for i in range(len(remaining_ints)):
        integer = remaining_ints[i]
        next_remaining = remaining_ints[i+1:]
        subset_sum(partial_sum + [integer], next_remaining, target)
        
    
    
```

```python
subset_sum([], [1, 2, 3, 4, 5, 6, 7], 6)
print(solutions)
```
