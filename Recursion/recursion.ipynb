```python
#base case
memo = {0:1,1:1,2:2,3:4}
def colored_blocks(n):
    global memo
    if n in memo:
        return memo[n]
    memo[n] = colored_blocks(n-1)+colored_blocks(n-2)+colored_blocks(n-3)+colored_blocks(n-4)
    #print(memo)
    return memo[n] 
    
```

```python
colored_blocks(50)
```

```python
memo2 = {}
def stirling2(n,k):
    global memo2
    
    for i in range(1, n+1):
        if (i,1) not in memo2:
            memo2[(i,1)] = 1
        if (i,i) not in memo2:
            memo2[(i,i)] = 1
        

    if k > n or n < 1 or k < 1:
        return 0
        
    elif (n,k) in memo2:
        return memo2[(n,k)]
    memo2[(n,k)] = stirling2(n-1,k-1) + k * stirling2(n-1,k)

    return memo2[(n,k)]



    
```

```python
print(stirling2(10, 5))
```
